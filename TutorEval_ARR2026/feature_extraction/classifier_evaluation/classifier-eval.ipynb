```python
import pandas as pd
import numpy as np
from sklearn.metrics import roc_auc_score
```

```python
feature_files = {
    'Politeness':"./Politeness.csv",
    'Agency':"./Agency.csv",
    'PressAccuracy':"./PressAccuracy.csv",
    'Uptake':"./Uptake.csv",
    'PressReasoning':"./PressReasoning.csv"
}
```

```python
human_auc_summary = {}
for rubric, file_path in feature_files.items():
    print(f"EVALUATING RUBRIC: {rubric.upper()}")
    df = pd.read_csv(file_path, encoding='cp1252')

    human_df = df.dropna(subset=['human', 'classifier_probability']).copy()
    human_classes = human_df['human'].unique()
    
    auc_human = roc_auc_score(human_df['human'], human_df['classifier_probability'])
    print(f"AUC (Human as Ground Truth): {auc_human:.3f}")
    human_auc_summary[rubric] = round(auc_human, 3)

    print()
    comparison_df = pd.DataFrame({
        'Rubric': list(human_auc_summary.keys()),
        'AUC (Human GT)': [human_auc_summary.get(r, np.nan) for r in human_auc_summary.keys()],

})
comparison_df.to_csv("auc_comparison_extreme_human.csv", index=False)

```
